# Numerical Analysis Sample

This notebook approximates the root of $f(x) = x^2 - 2$ using the bisection method.

```python
def bisection(function, left, right, tolerance=1e-6, max_iterations=100):
    """Approximate a root of function on [left, right]."""
    if function(left) * function(right) > 0:
        raise ValueError("The interval must contain a sign change.")

    for _ in range(max_iterations):
        midpoint = (left + right) / 2
        if abs(function(midpoint)) < tolerance or (right - left) / 2 < tolerance:
            return midpoint

        if function(left) * function(midpoint) < 0:
            right = midpoint
        else:
            left = midpoint

    return (left + right) / 2
```

```python
root = bisection(lambda x: x**2 - 2, 1, 2)
print(f"Approximate root: {root:.6f}")
print(f"Verification: root² - 2 = {root**2 - 2:.2e}")
```
